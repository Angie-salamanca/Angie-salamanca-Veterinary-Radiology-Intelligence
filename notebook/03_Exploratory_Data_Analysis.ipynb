# ==========================================
# Exploratory Data Analysis (EDA)
# ==========================================

## Objective

The objective of this notebook is to explore the cleaned veterinary radiology dataset in order to identify meaningful patterns, trends and relationships.

The analysis focuses on answering practical business and clinical questions such as:

- Which species are examined most frequently?
- Which breeds are the most common?
- What is the distribution of patient age and weight?
- How many radiographs are usually performed?
- How often are medical reports generated?
- Which veterinarians refer the largest number of cases?
- How does radiology workload change throughout the year?

The insights obtained in this notebook will later support the development of dashboards and business intelligence reports.

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "browser"

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
# ==========================================
# LOAD CLEAN DATASET
# ==========================================

radiology = pd.read_csv(
    "../data/processed/radiology_clean.csv",
    parse_dates=["exam_date", "birth_date"]
)

print("Dataset loaded successfully!")
```

# ==========================================
# Species Analysis
# ==========================================

The first step is to understand which animal species are most frequently referred for radiographic examination.

This analysis helps identify the primary patient population served by the radiology service.

```python
# ==========================================
# SPECIES DISTRIBUTION
# ==========================================
species_counts = (
    radiology["species"]
    .value_counts()
)

species_counts
```

```python
# ==========================================
# MISSING SPECIES VALUES
# ==========================================

missing_species = radiology["species"].isna().sum()

print("Missing species values:", missing_species)
```

```python
# ==========================================
# SPECIES DISTRIBUTION - FINAL TABLE
# ==========================================

species_table = (
    radiology["species"]
    .value_counts(dropna=False)
    .rename_axis("Species")
    .reset_index(name="Cases")
)

species_table["Percentage"] = (
    species_table["Cases"] / len(radiology) * 100
).round(2)

species_table
```

```python
# ==========================================
# SPECIES DISTRIBUTION - FINAL CHART
# ==========================================

species_chart = species_table.copy()

species_chart["Species"] = (
    species_chart["Species"]
    .fillna("Not reported")
)

fig = px.bar(
    species_chart,
    x="Species",
    y="Cases",
    text="Cases",
    title="Species Distribution of Radiology Patients",
    labels={
        "Species": "Species",
        "Cases": "Number of Cases"
    }
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    xaxis_title="Species",
    yaxis_title="Number of Cases",
    title_x=0.5,
    template="plotly_white"
)

fig.show(renderer="browser")
```

### Species Distribution

The radiology service is primarily used for canine and feline patients.

Canine patients represented the majority of radiology records, with 496 cases (71.57%), followed by feline patients with 193 cases (27.85%).

Only one rabbit case (0.14%) was recorded during the study period, so this observation should not be interpreted as representative of the rabbit population.

Species information was missing in 3 records (0.43%).

```python
# ==========================================
# BREED DISTRIBUTION
# ==========================================

breed_counts = (
    radiology["breed"]
    .value_counts(dropna=False)
)

breed_counts.head(20)
```

```python
# ==========================================
# MISSING BREED VALUES
# ==========================================

missing_breed = radiology["breed"].isna().sum()

print("Missing breed values:", missing_breed)
```

```python
# ==========================================
# TOP 15 BREEDS
# ==========================================

top_breeds = (
    breed_counts
    .dropna()
    .head(15)
    .sort_values(ascending=True)
    .reset_index()
)

top_breeds.columns = [
    "Breed",
    "Cases"
]

top_breeds
```

```python
# ==========================================
# TOP 15 BREEDS - FINAL CHART
# ==========================================

fig = px.bar(
    top_breeds,
    x="Cases",
    y="Breed",
    orientation="h",
    text="Cases",
    title="Top 15 Breeds by Number of Radiology Cases",
    labels={
        "Breed": "Breed",
        "Cases": "Number of Cases"
    }
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    xaxis_title="Number of Cases",
    yaxis_title="Breed",
    title_x=0.5
)

fig.show(renderer="browser")
```

### Breed Distribution

The dataset contains a wide variety of recorded breeds. CRIOLLO was the most frequently recorded breed, with 256 radiology cases, followed by FRENCH POODLE with 63 cases and MESTIZO with 41 cases.

The remaining top breeds were recorded less frequently, showing that radiology demand was distributed across a wide range of breed categories.

Breed information was missing in 7 records (1.0% of the dataset).

```python
# ==========================================
# SEX DISTRIBUTION WITHIN EACH SPECIES
# ==========================================

sex_by_species_percentage = (
    pd.crosstab(
        radiology["species"],
        radiology["sex"],
        normalize="index"
    ) * 100
)

sex_by_species_percentage = (
    sex_by_species_percentage
    .round(2)
)

sex_by_species_percentage
```

```python
# ==========================================
# SEX DISTRIBUTION WITHIN EACH SPECIES - FINAL CHART
# ==========================================

sex_species_chart = (
    sex_by_species_percentage
    .reset_index()
    .melt(
        id_vars="species",
        var_name="Sex",
        value_name="Percentage"
    )
)

fig = px.bar(
    sex_species_chart,
    x="species",
    y="Percentage",
    color="Sex",
    barmode="group",
    text="Percentage",
    title="Sex Distribution Within Each Species",
    labels={
        "species": "Species",
        "Percentage": "Percentage (%)",
        "Sex": "Sex"
    }
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside",
    hovertemplate=(
        "Species=%{x}<br>"
        "Sex=%{fullData.name}<br>"
        "Percentage=%{y:.1f}%"
        "<extra></extra>"
    )
)

fig.update_layout(
    yaxis_title="Percentage (%)",
    xaxis_title="Species",
    yaxis_range=[0, 100],
    title_x=0.5
)

fig.show(renderer="browser")
```

### Sex Distribution Within Each Species

The sex distribution was relatively balanced between dogs and cats.

Among canine patients, 54.8% were recorded as H and 45.2% as M. Among feline patients, 54.5% were recorded as H and 45.5% as M.

The rabbit category contains only one patient, so its 100% H proportion should not be interpreted as representative of the rabbit population.

> **Data quality note:** 10 records did not contain sex information.

```python
# ==========================================
# PATIENT AGE AT EXAMINATION
# ==========================================

radiology["patient_age_years"] = (
    radiology["exam_date"] - radiology["birth_date"]
).dt.days / 365.25
```

```python
# ==========================================
# PATIENT AGE DISTRIBUTION
# ==========================================

radiology["patient_age_years"].describe()
```

```python
# ==========================================
# MISSING PATIENT AGE VALUES
# ==========================================

missing_age = radiology["patient_age_years"].isna().sum()

print(
    "Missing patient age values:",
    missing_age
)
```

```python
# ==========================================
# PATIENTS YOUNGER THAN 1 YEAR
# ==========================================

young_patients_count = (
    radiology["patient_age_years"] < 1
).sum()

print(
    "Patients younger than 1 year:",
    young_patients_count
)
```

```python
# ==========================================
# PATIENT AGE DISTRIBUTION - FINAL CHART
# ==========================================

age_data = (
    radiology[
        radiology["patient_age_years"].notna()
    ]
)

mean_age = age_data["patient_age_years"].mean()
median_age = age_data["patient_age_years"].median()

fig = px.histogram(
    age_data,
    x="patient_age_years",
    nbins=12,
    title="Distribution of Patient Age at Examination",
    labels={
        "patient_age_years": "Age (years)",
        "count": "Number of Patients"
    },
    text_auto=True
)

fig.add_vline(
    x=mean_age,
    line_dash="dash",
    annotation_text=f"Mean: {mean_age:.2f} years",
    annotation_position="top"
)

fig.add_vline(
    x=median_age,
    line_dash="dot",
    annotation_text=f"Median: {median_age:.2f} years",
    annotation_position="top"
)

fig.update_layout(
    xaxis_title="Age (years)",
    yaxis_title="Number of Patients",
    title_x=0.5
)

fig.show(renderer="browser")
```

### Patient Age Distribution

The analysis included 531 radiology records with available patient age information.

The mean patient age was 6.77 years, while the median age was 6.54 years. Patients ranged from 0 to 24.62 years of age.

A total of 71 patients were younger than one year at the time of examination.

Age information was missing for 162 records, which should be considered when interpreting the age distribution.

```python
# ==========================================
# AGE DISTRIBUTION BY SPECIES
# ==========================================

age_by_species = (
    radiology
    .groupby("species")["patient_age_years"]
    .agg(
        Cases="count",
        Mean="mean",
        Median="median",
        Minimum="min",
        Maximum="max"
    )
    .round(2)
)

age_by_species
```

## Medical Report Status

This section analyzes the presence of a medical report in the radiology records.

The medical report variable indicates whether a formal radiology report was recorded for the examination. Missing values are treated separately because a missing entry means that the report status was not recorded.

```python
# ==========================================
# MEDICAL REPORT STATUS - FINAL TABLE
# ==========================================

medical_report_table = (
    radiology["medical_report"]
    .value_counts(dropna=False)
    .rename_axis("Medical report")
    .reset_index(name="Cases")
)

medical_report_table["Medical report"] = (
    medical_report_table["Medical report"]
    .fillna("Not reported")
)

medical_report_table["Percentage"] = (
    medical_report_table["Cases"]
    / len(radiology)
    * 100
).round(2)

medical_report_table
```

```python
# ==========================================
# MEDICAL REPORT STATUS - FINAL CHART
# ==========================================

fig = px.bar(
    medical_report_table,
    x="Medical report",
    y="Cases",
    text="Cases",
    title="Medical Report Status of Radiology Cases",
    labels={
        "Medical report": "Medical Report Status",
        "Cases": "Number of Cases"
    }
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    xaxis_title="Medical Report Status",
    yaxis_title="Number of Cases",
    title_x=0.5
)

fig.show()
```

### Interpretation

Medical reports were recorded in 301 of 693 radiology records (43.4%), while 350 records (50.5%) were marked as having no medical report.

An additional 42 records (6.1%) had missing report information.

Because the dataset records report status rather than the clinical reason for generating or not generating a report, these results should be interpreted as an operational description of report documentation.

```python
# ==========================================
# MEDICAL REPORT BY RADIOGRAPH COUNT
# ==========================================

report_by_radiograph_count = pd.crosstab(
    radiology["radiograph_count"],
    radiology["medical_report"]
)

report_by_radiograph_count
```

```python
# ==========================================
# MEDICAL REPORT PERCENTAGE BY RADIOGRAPH COUNT
# ==========================================

report_by_radiograph_percentage = (
    pd.crosstab(
        radiology["radiograph_count"],
        radiology["medical_report"],
        normalize="index"
    ) * 100
)

report_by_radiograph_percentage = (
    report_by_radiograph_percentage.round(2)
)

report_by_radiograph_percentage
```

### Medical Report Status by Number of Radiographs

A strong relationship was observed between the number of radiographs performed and the presence of a medical report.

Among examinations with one radiograph, only 2.78% had a medical report, while 97.22% did not.

For examinations with two radiographs, 76.54% had a medical report. This proportion increased to 82.35% among examinations with three radiographs.

All examinations with four, five or six radiographs had a medical report. However, these groups contained very few cases and should therefore be interpreted cautiously.

Overall, examinations involving two or more radiographs were substantially more likely to have an associated medical report than examinations involving a single radiograph.

```python
# ==========================================
# STANDARDIZE REFERRING VETERINARIAN NAMES
# ==========================================

radiology["referring_veterinarian_clean"] = (
    radiology["referring_veterinarian"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.replace(r"\s+", " ", regex=True)
    .str.replace(r"\bDR\.", "DR", regex=True)
    .str.replace(r"\bVET\.", "VET", regex=True)
)
```

Ambiguous names and abbreviations were intentionally kept separate when there was not enough evidence to assign them reliably to a specific veterinarian.

```python
# ==========================================
# VETERINARIANS WITH AT LEAST 5 CASES
# ==========================================

veterinarian_case_counts = (
    radiology["referring_veterinarian_clean"]
    .value_counts()
)

veterinarians_5plus = (
    veterinarian_case_counts[
        veterinarian_case_counts >= 5
    ]
    .index
)

print(
    "Number of veterinarians with at least 5 cases:",
    len(veterinarians_5plus)
)
```

```python
# ==========================================
# FINAL REPORT METRICS BY VETERINARIAN
# ==========================================

veterinarian_report_analysis = (
    pd.DataFrame({
        "Total Cases": veterinarian_case_counts[
            veterinarian_case_counts.index.isin(
                veterinarians_5plus
            )
        ],
        "Reports": report_by_veterinarian_5plus.get("SI", 0),
        "No Reports": report_by_veterinarian_5plus.get("NO", 0)
    })
)

veterinarian_report_analysis["Missing Report"] = (
    veterinarian_report_analysis["Total Cases"]
    - veterinarian_report_analysis["Reports"]
    - veterinarian_report_analysis["No Reports"]
)

veterinarian_report_analysis["Report Percentage"] = (
    veterinarian_report_analysis["Reports"]
    / (
        veterinarian_report_analysis["Reports"]
        + veterinarian_report_analysis["No Reports"]
    )
    * 100
).round(2)

veterinarian_report_analysis = (
    veterinarian_report_analysis
    .sort_values("Total Cases", ascending=False)
)

veterinarian_report_analysis
```

```python
# ==========================================
# TOP 15 VETERINARIANS BY CASE VOLUME
# ==========================================

top15_veterinarian_report = (
    veterinarian_report_analysis
    .head(15)
    .sort_values("Report Percentage", ascending=True)
    .reset_index()
)

top15_veterinarian_report
```

- Medical report status: 42 records

### Referring Veterinarian Data Quality

Ambiguous names and abbreviations were kept separate when there was not enough evidence to assign them reliably to a specific veterinarian.

The dataset contains only three examination records from 2021, while the vast majority of records correspond to 2022.

### Small Categories

Some categories contain very few observations.

For example, the rabbit category contains only one patient. Percentages and distributions based on such small groups should not be generalized to the broader population.

### Missingness and Interpretation

Different analyses are based on different subsets of records because of missing values. Therefore, percentages and summary statistics should always be interpreted together with the number of observations available for each variable.

### Analytical Limitations

The dataset represents patients who received radiology services during the available study period. The findings therefore describe the population represented in this dataset and should not automatically be generalized to the broader veterinary population.
