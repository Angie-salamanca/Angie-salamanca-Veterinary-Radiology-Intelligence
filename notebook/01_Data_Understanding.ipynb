## Weight Assessment

### Domain Knowledge

Based on veterinary practice, some values have a specific clinical meaning:

- `NS` indicates that the patient's weight was **not recorded**, because it was not required during the examination.
- `NO` indicates that **no weight was determined** at the time of the examination.
- Decimal values written with commas follow the local numeric notation and represent valid weights.
- Values such as `26kg` include the measurement unit and require standardization.

These observations will guide the data cleaning process in the next notebook.

## Domain Knowledge

This project was developed using real veterinary clinical records.

The data cleaning strategy was not based only on programming techniques, but also on veterinary clinical knowledge to correctly interpret missing values, abbreviations, and data entry practices.

This approach improves data quality while preserving the clinical meaning of the records.
